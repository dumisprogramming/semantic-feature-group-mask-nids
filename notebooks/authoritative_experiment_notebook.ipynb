```python
import pandas as pd
from collections import Counter
from pathlib import Path
import time

def audit_labels(files, dataset_name, chunksize=500_000):
    dataset_counts = Counter()
    file_results = []

    print(f"\nAuditing {dataset_name}")
    print("=" * 70)

    for number, file in enumerate(files, start=1):
        start_time = time.time()

        # Find the real label-column name, including possible whitespace
        raw_columns = pd.read_csv(file, nrows=0).columns.tolist()

        label_candidates = [
            column for column in raw_columns
            if str(column).strip().lower() == "label"
        ]

        if not label_candidates:
            print(f"Label column not found: {file.name}")
            continue

        actual_label_column = label_candidates[0]
        file_counts = Counter()
        total_rows = 0

        for chunk in pd.read_csv(
            file,
            usecols=[actual_label_column],
            chunksize=chunksize,
            keep_default_na=False
        ):
            labels = (
                chunk[actual_label_column]
                .astype(str)
                .str.strip()
            )

            file_counts.update(labels)
            total_rows += len(chunk)

        dataset_counts.update(file_counts)
        elapsed = time.time() - start_time

        file_results.append({
            "dataset": dataset_name,
            "file": file.name,
            "rows": total_rows,
            "labels": dict(file_counts)
        })

        print(
            f"[{number}/{len(files)}] {file.name}\n"
            f"Rows: {total_rows:,} | "
            f"Labels: {dict(file_counts)} | "
            f"Time: {elapsed:.1f} seconds\n"
        )

    print(f"\n{dataset_name} TOTAL")
    print("Total rows:", f"{sum(dataset_counts.values()):,}")

    for label, count in dataset_counts.most_common():
        print(f"{label}: {count:,}")

    return dataset_counts, file_results
```

```python
import pandas as pd

def normalize_binary_label(series):
    cleaned = (
        series.astype(str)
        .str.strip()
    )

    # Repeated header rows and blank labels become invalid
    invalid = (
        cleaned.str.lower().isin(["label", "", "nan"])
    )

    binary = pd.Series(pd.NA, index=series.index, dtype="Int8")

    # Benign spelling differs only by capitalization
    binary.loc[cleaned.str.lower() == "benign"] = 0

    # Every valid non-benign label is an attack
    binary.loc[~invalid & (cleaned.str.lower() != "benign")] = 1

    return binary
```

```python
def summarize_binary_counts(label_counts):
    benign = sum(
        count for label, count in label_counts.items()
        if label.strip().lower() == "benign"
    )

    invalid = sum(
        count for label, count in label_counts.items()
        if label.strip().lower() in ["label", "", "nan"]
    )

    total = sum(label_counts.values())
    valid = total - invalid
    attack = valid - benign

    return {
        "original_rows": total,
        "invalid_rows": invalid,
        "valid_rows": valid,
        "benign": benign,
        "attack": attack,
        "attack_percentage": round(attack / valid * 100, 2)
    }

print("CICIDS2017:")
print(summarize_binary_counts(counts_2017))

print("\nCSE-CIC-IDS2018:")
print(summarize_binary_counts(counts_2018))
```

```python
import pandas as pd
import numpy as np
import time

predictor_columns = sorted(set(common_columns) - {"Label"})

def standardize_chunk(chunk, dataset):
    # Remove whitespace from column names
    chunk.columns = [
        str(column).strip()
        for column in chunk.columns
    ]

    if dataset == "2017":
        chunk = chunk.drop(
            columns=["Fwd Header Length.1"],
            errors="ignore"
        )

    elif dataset == "2018":
        chunk = chunk.drop(
            columns=[
                "Protocol",
                "Timestamp",
                "Flow ID",
                "Src IP",
                "Dst IP",
                "Src Port"
            ],
            errors="ignore"
        )

        chunk = chunk.rename(
            columns=mapping_2018_to_2017
        )

    return chunk


def audit_predictors(files, dataset, chunksize=100_000):
    missing_counts = pd.Series(
        0, index=predictor_columns, dtype="int64"
    )
    nonnumeric_counts = missing_counts.copy()
    positive_inf_counts = missing_counts.copy()
    negative_inf_counts = missing_counts.copy()

    total_original = 0
    total_valid = 0
    total_invalid_labels = 0

    print(f"\nAuditing predictors: {dataset}")
    print("=" * 70)

    for file_number, file in enumerate(files, start=1):
        start_time = time.time()
        file_original = 0
        file_valid = 0
        file_invalid = 0

        for chunk in pd.read_csv(
            file,
            chunksize=chunksize,
            low_memory=False,
            encoding_errors="replace"
        ):
            chunk = standardize_chunk(chunk, dataset)

            labels = (
                chunk["Label"]
                .astype(str)
                .str.strip()
                .str.lower()
            )

            invalid_label = labels.isin(
                ["label", "", "nan"]
            )

            file_original += len(chunk)
            file_invalid += int(invalid_label.sum())

            # Remove invalid/repeated-header rows
            chunk = chunk.loc[
                ~invalid_label,
                predictor_columns
            ]

            file_valid += len(chunk)

            raw_missing = chunk.isna()

            numeric = chunk.apply(
                pd.to_numeric,
                errors="coerce"
            )

            converted_missing = numeric.isna()

            missing_counts += raw_missing.sum().astype("int64")

            # Values present originally but not convertible to numbers
            nonnumeric_counts += (
                converted_missing & ~raw_missing
            ).sum().astype("int64")

            positive_inf_counts += (
                numeric == np.inf
            ).sum().astype("int64")

            negative_inf_counts += (
                numeric == -np.inf
            ).sum().astype("int64")

            del chunk, numeric, raw_missing, converted_missing

        total_original += file_original
        total_valid += file_valid
        total_invalid_labels += file_invalid

        elapsed = time.time() - start_time

        print(
            f"[{file_number}/{len(files)}] {file.name}\n"
            f"Original: {file_original:,} | "
            f"Valid: {file_valid:,} | "
            f"Invalid labels: {file_invalid:,} | "
            f"Time: {elapsed:.1f} seconds\n"
        )

    report = pd.DataFrame({
        "missing": missing_counts,
        "nonnumeric": nonnumeric_counts,
        "positive_infinity": positive_inf_counts,
        "negative_infinity": negative_inf_counts
    })

    report["total_issues"] = report.sum(axis=1)
    report = report.sort_values(
        "total_issues",
        ascending=False
    )

    print("\nSummary")
    print("Original rows:", f"{total_original:,}")
    print("Invalid-label rows:", f"{total_invalid_labels:,}")
    print("Valid rows:", f"{total_valid:,}")
    print(
        "Features containing issues:",
        int((report["total_issues"] > 0).sum())
    )

    print("\nFeatures with detected issues:")
    display(report[report["total_issues"] > 0])

    return report
```

```python
from collections import Counter
import pandas as pd
import numpy as np

def audit_invalid_rate_labels(files, dataset, chunksize=250_000):
    affected_labels = Counter()
    valid_rows = 0
    affected_rows = 0

    for number, file in enumerate(files, start=1):
        raw_header = pd.read_csv(file, nrows=0).columns.tolist()
        header_lookup = {
            str(column).strip(): column
            for column in raw_header
        }

        if dataset == "2017":
            byte_column = header_lookup["Flow Bytes/s"]
            packet_column = header_lookup["Flow Packets/s"]
        else:
            byte_column = header_lookup["Flow Byts/s"]
            packet_column = header_lookup["Flow Pkts/s"]

        label_column = header_lookup["Label"]

        for chunk in pd.read_csv(
            file,
            usecols=[byte_column, packet_column, label_column],
            chunksize=chunksize,
            low_memory=False
        ):
            labels = (
                chunk[label_column]
                .astype(str)
                .str.strip()
            )

            valid_label = ~labels.str.lower().isin(
                ["label", "", "nan"]
            )

            flow_bytes = pd.to_numeric(
                chunk[byte_column],
                errors="coerce"
            )

            flow_packets = pd.to_numeric(
                chunk[packet_column],
                errors="coerce"
            )

            affected = valid_label & (
                ~np.isfinite(flow_bytes) |
                ~np.isfinite(flow_packets)
            )

            valid_rows += int(valid_label.sum())
            affected_rows += int(affected.sum())
            affected_labels.update(labels[affected])

        print(f"[{number}/{len(files)}] {file.name}")

    benign_affected = sum(
        count for label, count in affected_labels.items()
        if label.lower() == "benign"
    )

    attack_affected = affected_rows - benign_affected

    print(f"\nDataset: {dataset}")
    print(f"Valid rows: {valid_rows:,}")
    print(f"Affected rows: {affected_rows:,}")
    print(f"Affected percentage: {affected_rows / valid_rows * 100:.4f}%")
    print(f"Affected benign: {benign_affected:,}")
    print(f"Affected attacks: {attack_affected:,}")

    print("\nAffected rows by original label:")
    for label, count in affected_labels.most_common():
        print(f"{label}: {count:,}")

    return affected_labels
```

```python
import pandas as pd
import numpy as np

def clean_chunk(chunk, dataset):
    chunk = standardize_chunk(chunk, dataset)

    labels = (
        chunk["Label"]
        .astype(str)
        .str.strip()
    )

    valid_label = ~labels.str.lower().isin(
        ["label", "", "nan"]
    )

    chunk = chunk.loc[valid_label].copy()
    labels = labels.loc[valid_label]

    # Binary target
    chunk["target"] = (
        labels.str.lower() != "benign"
    ).astype("int8")

    # Convert all predictors to numeric
    chunk[predictor_columns] = chunk[
        predictor_columns
    ].apply(pd.to_numeric, errors="coerce")

    # Undefined rates become missing—not zero
    chunk[predictor_columns] = chunk[
        predictor_columns
    ].replace([np.inf, -np.inf], np.nan)

    return chunk[predictor_columns + ["target"]]
```

```python
import pandas as pd
import numpy as np
import time

def create_uniform_sample(
    files,
    dataset,
    valid_population,
    desired_rows=1_000_000,
    seed=2026,
    chunksize=100_000
):
    sampling_probability = min(
        1.0,
        desired_rows / valid_population
    )

    rng = np.random.default_rng(seed)
    samples = []

    print(f"\nSampling {dataset}")
    print("Sampling probability:", sampling_probability)

    for file_number, file in enumerate(files, start=1):
        start = time.time()
        selected_from_file = 0

        for raw_chunk in pd.read_csv(
            file,
            chunksize=chunksize,
            low_memory=False,
            encoding_errors="replace"
        ):
            cleaned = clean_chunk(raw_chunk, dataset)

            selection = (
                rng.random(len(cleaned))
                < sampling_probability
            )

            selected = cleaned.loc[selection].copy()

            selected[predictor_columns] = selected[
                predictor_columns
            ].astype("float32")

            selected["target"] = selected[
                "target"
            ].astype("int8")

            samples.append(selected)
            selected_from_file += len(selected)

        print(
            f"[{file_number}/{len(files)}] {file.name} | "
            f"Selected: {selected_from_file:,} | "
            f"Time: {time.time() - start:.1f}s"
        )

    sample = pd.concat(
        samples,
        ignore_index=True
    )

    del samples

    print("\nFinal sampled rows:", f"{len(sample):,}")
    print("Benign:", f"{(sample['target'] == 0).sum():,}")
    print("Attack:", f"{(sample['target'] == 1).sum():,}")
    print(
        "Attack rate:",
        f"{sample['target'].mean() * 100:.2f}%"
    )

    return sample
```

```python
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    matthews_corrcoef,
    confusion_matrix
)
import pandas as pd

def evaluate_binary_model(model, X, y, threshold):
    probability = model.predict_proba(X)[:, 1]
    prediction = (probability >= threshold).astype("int8")

    tn, fp, fn, tp = confusion_matrix(
        y,
        prediction,
        labels=[0, 1]
    ).ravel()

    return {
        "rows": len(y),
        "accuracy": accuracy_score(y, prediction),
        "balanced_accuracy": balanced_accuracy_score(y, prediction),
        "precision": precision_score(
            y, prediction, zero_division=0
        ),
        "recall": recall_score(
            y, prediction, zero_division=0
        ),
        "f1": f1_score(
            y, prediction, zero_division=0
        ),
        "roc_auc": roc_auc_score(y, probability),
        "pr_auc": average_precision_score(y, probability),
        "mcc": matthews_corrcoef(y, prediction),
        "false_positive_rate": fp / (fp + tn),
        "false_negative_rate": fn / (fn + tp),
        "tn": int(tn),
        "fp": int(fp),
        "fn": int(fn),
        "tp": int(tp)
    }

baseline_results = pd.DataFrame({
    "Validation": evaluate_binary_model(
        baseline_model,
        X_validation_imputed,
        y_validation,
        selected_threshold
    ),
    "Source_Test": evaluate_binary_model(
        baseline_model,
        X_source_test_imputed,
        y_source_test,
        selected_threshold
    )
}).T

display(
    baseline_results.round(6)
)
```

```python
from itertools import combinations
import pandas as pd
import numpy as np
import time

feature_index = {
    feature: position
    for position, feature in enumerate(predictor_columns)
}

group_indices = {
    group: [
        feature_index[feature]
        for feature in features
    ]
    for group, features in semantic_groups.items()
}

training_medians = np.asarray(
    imputer.statistics_,
    dtype=np.float32
)
```

Create masked versions:

```python
def apply_group_loss(
    X,
    missing_groups,
    replacement
):
    X_masked = np.array(
        X,
        dtype=np.float32,
        copy=True
    )

    columns_to_mask = []

    for group in missing_groups:
        columns_to_mask.extend(
            group_indices[group]
        )

    columns_to_mask = sorted(
        set(columns_to_mask)
    )

    if replacement == "zero":
        X_masked[:, columns_to_mask] = 0.0

    elif replacement == "median":
        X_masked[:, columns_to_mask] = (
            training_medians[columns_to_mask]
        )

    else:
        raise ValueError(
            "Replacement must be zero or median."
        )

    return X_masked
```

Run complete, single-group and unseen pairwise-group tests:

```python
stress_results = []

# Complete-feature reference
complete_result = evaluate_binary_model(
    baseline_model,
    X_source_test_imputed,
    y_source_test,
    selected_threshold
)

complete_result.update({
    "scenario": "Complete",
    "loss_type": "none",
    "missing_groups": "None",
    "replacement": "none",
    "features_removed": 0
})

stress_results.append(complete_result)

group_names = list(semantic_groups.keys())

scenarios = []

# Single-group loss
for group in group_names:
    scenarios.append(
        ("single", (group,))
    )

# Pairwise feature-group loss
for pair in combinations(group_names, 2):
    scenarios.append(
        ("pairwise_unseen", pair)
    )

for loss_type, missing_groups in scenarios:
    number_removed = sum(
        len(semantic_groups[group])
        for group in missing_groups
    )

    for replacement in ["zero", "median"]:
        start = time.time()

        X_masked = apply_group_loss(
            X_source_test_imputed,
            missing_groups,
            replacement
        )

        result = evaluate_binary_model(
            baseline_model,
            X_masked,
            y_source_test,
            selected_threshold
        )

        result.update({
            "scenario": "+".join(missing_groups),
            "loss_type": loss_type,
            "missing_groups": "+".join(missing_groups),
            "replacement": replacement,
            "features_removed": number_removed
        })

        stress_results.append(result)

        print(
            loss_type,
            "|",
            "+".join(missing_groups),
            "|",
            replacement,
            "| F1:",
            round(result["f1"], 4),
            "| Time:",
            round(time.time() - start, 1),
            "seconds"
        )

        del X_masked
```

Next: train the group-aware model

The model will see:

Complete flows
Single-group loss during training
Four group-availability indicators
Median replacement learned only from source training
No pairwise group loss during training
No target data

```python
group_names = list(semantic_groups.keys())

availability_names = [
    f"available__{group}"
    for group in group_names
]

group_aware_feature_names = (
    predictor_columns + availability_names
)

print("Group-aware predictors:", len(group_aware_feature_names))
```

Create complete plus randomly assigned single-group-loss training examples:

```python
def create_group_aware_training_data(
    X,
    y,
    seed=2026
):
    X = np.asarray(X, dtype=np.float32)
    y = np.asarray(y, dtype=np.int8)

    number_rows = len(X)
    number_groups = len(group_names)

    rng = np.random.default_rng(seed)

    # Every row gets one randomly selected missing group
    assigned_groups = rng.integers(
        0,
        number_groups,
        size=number_rows
    )

    X_masked = X.copy()

    masked_availability = np.ones(
        (number_rows, number_groups),
        dtype=np.float32
    )

    for group_number, group in enumerate(group_names):
        rows = np.where(
            assigned_groups == group_number
        )[0]

        columns = group_indices[group]

        X_masked[np.ix_(rows, columns)] = (
            training_medians[columns]
        )

        masked_availability[
            rows,
            group_number
        ] = 0.0

        print(
            group,
            "masked training rows:",
            f"{len(rows):,}"
        )

    complete_availability = np.ones(
        (number_rows, number_groups),
        dtype=np.float32
    )

    X_complete_augmented = np.hstack([
        X,
        complete_availability
    ])

    X_masked_augmented = np.hstack([
        X_masked,
        masked_availability
    ])

    X_augmented = np.vstack([
        X_complete_augmented,
        X_masked_augmented
    ])

    y_augmented = np.concatenate([
        y,
        y
    ])

    return X_augmented, y_augmented
```

Create validation data containing complete and all four known single-loss conditions:

```python
def create_known_condition_validation_data(X, y):
    X = np.asarray(X, dtype=np.float32)
    y = np.asarray(y, dtype=np.int8)

    number_rows = len(X)
    number_groups = len(group_names)

    X_versions = []
    y_versions = []

    # Complete condition
    complete_indicators = np.ones(
        (number_rows, number_groups),
        dtype=np.float32
    )

    X_versions.append(
        np.hstack([X, complete_indicators])
    )
    y_versions.append(y)

    # Four single-group-loss conditions
    for group_number, group in enumerate(group_names):
        X_masked = X.copy()

        columns = group_indices[group]
        X_masked[:, columns] = training_medians[columns]

        indicators = np.ones(
            (number_rows, number_groups),
            dtype=np.float32
        )
        indicators[:, group_number] = 0.0

        X_versions.append(
            np.hstack([X_masked, indicators])
        )
        y_versions.append(y)

    return (
        np.vstack(X_versions),
        np.concatenate(y_versions)
    )
```

Build the augmented arrays:

```python
X_train_group_aware, y_train_group_aware = (
    create_group_aware_training_data(
        X_train_imputed,
        y_train,
        seed=2026
    )
)

X_validation_group_aware, y_validation_group_aware = (
    create_known_condition_validation_data(
        X_validation_imputed,
        y_validation
    )
)

print(
    "Augmented training shape:",
    X_train_group_aware.shape
)

print(
    "Known-condition validation shape:",
    X_validation_group_aware.shape
)
```

Train using the same main hyperparameters as the baseline:

```python
group_aware_model = xgb.XGBClassifier(
    objective="binary:logistic",
    n_estimators=1200,
    learning_rate=0.05,
    max_depth=6,
    min_child_weight=5,
    subsample=0.80,
    colsample_bytree=0.80,
    reg_alpha=0.01,
    reg_lambda=1.0,
    tree_method="hist",
    device=device,
    eval_metric="logloss",
    early_stopping_rounds=50,
    random_state=2026,
    n_jobs=-1
)

start_time = time.time()

group_aware_model.fit(
    X_train_group_aware,
    y_train_group_aware,
    eval_set=[
        (
            X_validation_group_aware,
            y_validation_group_aware
        )
    ],
    verbose=50
)

group_aware_training_time = (
    time.time() - start_time
)

print("\nTraining completed")
print(
    "Training time:",
    round(group_aware_training_time, 2),
    "seconds"
)
print(
    "Best iteration:",
    group_aware_model.best_iteration
)
print(
    "Best validation logloss:",
    group_aware_model.best_score
)
```

4. Fit Platt calibration

```python
from sklearn.linear_model import LogisticRegression

def probability_to_logit(probability):
    probability = np.clip(
        probability,
        1e-7,
        1 - 1e-7
    )

    return np.log(
        probability / (1 - probability)
    ).reshape(-1, 1)


raw_calibrator_probability = (
    group_aware_model.predict_proba(
        X_calibrator_known
    )[:, 1]
)

platt_calibrator = LogisticRegression(
    C=1_000_000,
    solver="lbfgs",
    max_iter=1000,
    random_state=2026
)

platt_calibrator.fit(
    probability_to_logit(
        raw_calibrator_probability
    ),
    y_calibrator_known
)

print(
    "Platt coefficient:",
    platt_calibrator.coef_[0, 0]
)

print(
    "Platt intercept:",
    platt_calibrator.intercept_[0]
)
```

5. Compare raw and calibrated probabilities

```python
from sklearn.metrics import (
    brier_score_loss,
    log_loss
)

def expected_calibration_error(
    y_true,
    probability,
    bins=15
):
    edges = np.linspace(0, 1, bins + 1)
    total = len(y_true)
    ece = 0.0

    for lower, upper in zip(
        edges[:-1],
        edges[1:]
    ):
        if upper == 1:
            selected = (
                (probability >= lower) &
                (probability <= upper)
            )
        else:
            selected = (
                (probability >= lower) &
                (probability < upper)
            )

        count = selected.sum()

        if count == 0:
            continue

        accuracy = np.mean(
            np.asarray(y_true)[selected]
        )

        confidence = np.mean(
            probability[selected]
        )

        ece += (
            count / total
        ) * abs(accuracy - confidence)

    return ece


raw_policy_probability = (
    group_aware_model.predict_proba(
        X_policy_known
    )[:, 1]
)

calibrated_policy_probability = (
    platt_calibrator.predict_proba(
        probability_to_logit(
            raw_policy_probability
        )
    )[:, 1]
)

calibration_results = pd.DataFrame({
    "Raw": {
        "Brier": brier_score_loss(
            y_policy_known,
            raw_policy_probability
        ),
        "LogLoss": log_loss(
            y_policy_known,
            raw_policy_probability
        ),
        "ECE": expected_calibration_error(
            y_policy_known,
            raw_policy_probability
        )
    },
    "Platt_Calibrated": {
        "Brier": brier_score_loss(
            y_policy_known,
            calibrated_policy_probability
        ),
        "LogLoss": log_loss(
            y_policy_known,
            calibrated_policy_probability
        ),
        "ECE": expected_calibration_error(
            y_policy_known,
            calibrated_policy_probability
        )
    }
}).T

display(calibration_results.round(8))
```

2. Define selective metrics

```python
def calculate_aurc(
    y_true,
    probability,
    classification_threshold
):
    y_true = np.asarray(y_true)

    prediction = (
        probability >= classification_threshold
    ).astype("int8")

    confidence = np.maximum(
        probability,
        1 - probability
    )

    errors = (
        prediction != y_true
    ).astype("float64")

    order = np.argsort(-confidence)
    ordered_errors = errors[order]

    cumulative_errors = np.cumsum(
        ordered_errors
    )

    coverage_count = np.arange(
        1,
        len(y_true) + 1
    )

    selective_risk = (
        cumulative_errors / coverage_count
    )

    return float(np.mean(selective_risk))


def evaluate_selective_policy(
    y_true,
    probability,
    classification_threshold,
    confidence_cutoff
):
    y_true = np.asarray(y_true)

    prediction = (
        probability >= classification_threshold
    ).astype("int8")

    confidence = np.maximum(
        probability,
        1 - probability
    )

    accepted = (
        confidence >= confidence_cutoff
    )

    accepted_y = y_true[accepted]
    accepted_prediction = prediction[accepted]

    total_attacks = int(
        (y_true == 1).sum()
    )

    abstained_attacks = int(
        ((~accepted) & (y_true == 1)).sum()
    )

    accepted_false_negatives = int(
        (
            accepted &
            (y_true == 1) &
            (prediction == 0)
        ).sum()
    )

    if accepted.sum() == 0:
        return {
            "actual_coverage": 0.0,
            "selective_risk": np.nan,
            "accepted_accuracy": np.nan,
            "accepted_precision": np.nan,
            "accepted_recall": np.nan,
            "accepted_f1": np.nan
        }

    accepted_error = (
        accepted_prediction != accepted_y
    )

    return {
        "actual_coverage": float(
            accepted.mean()
        ),
        "accepted_rows": int(
            accepted.sum()
        ),
        "abstained_rows": int(
            (~accepted).sum()
        ),
        "selective_risk": float(
            accepted_error.mean()
        ),
        "accepted_accuracy": accuracy_score(
            accepted_y,
            accepted_prediction
        ),
        "accepted_precision": precision_score(
            accepted_y,
            accepted_prediction,
            zero_division=0
        ),
        "accepted_recall": recall_score(
            accepted_y,
            accepted_prediction,
            zero_division=0
        ),
        "accepted_f1": f1_score(
            accepted_y,
            accepted_prediction,
            zero_division=0
        ),
        "attack_abstention_rate": (
            abstained_attacks / total_attacks
        ),
        "accepted_false_negative_rate": (
            accepted_false_negatives /
            total_attacks
        )
    }
```

```python
source_selective_results = []

for missing_groups in combinations(
    group_names,
    2
):
    X_condition = create_group_aware_condition(
        X_source_test_imputed,
        missing_groups
    )

    raw_probability = (
        group_aware_model.predict_proba(
            X_condition
        )[:, 1]
    )

    calibrated_probability = (
        platt_calibrator.predict_proba(
            probability_to_logit(
                raw_probability
            )
        )[:, 1]
    )

    aurc = calculate_aurc(
        y_source_test,
        calibrated_probability,
        group_aware_threshold
    )

    for desired_coverage, cutoff in (
        confidence_cutoffs.items()
    ):
        result = evaluate_selective_policy(
            y_source_test,
            calibrated_probability,
            group_aware_threshold,
            cutoff
        )

        result.update({
            "missing_groups":
                "+".join(missing_groups),
            "desired_policy_coverage":
                desired_coverage,
            "source_confidence_cutoff":
                cutoff,
            "aurc": aurc
        })

        source_selective_results.append(
            result
        )

    del X_condition
```

1. Define the required functions

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    precision_recall_curve,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
from itertools import combinations
import numpy as np
import pandas as pd
import joblib
import json

desired_coverages = [0.80, 0.90, 0.95]

def probability_to_logit(probability):
    probability = np.clip(
        probability,
        1e-7,
        1 - 1e-7
    )

    return np.log(
        probability / (1 - probability)
    ).reshape(-1, 1)


def apply_group_loss(
    X,
    missing_groups,
    replacement="median"
):
    X_masked = np.asarray(
        X,
        dtype=np.float32
    ).copy()

    columns_to_mask = []

    for group in missing_groups:
        columns_to_mask.extend(
            group_indices[group]
        )

    columns_to_mask = sorted(
        set(columns_to_mask)
    )

    if replacement == "median":
        X_masked[:, columns_to_mask] = (
            training_medians[columns_to_mask]
        )
    elif replacement == "zero":
        X_masked[:, columns_to_mask] = 0.0
    else:
        raise ValueError(
            "Use 'median' or 'zero'."
        )

    return X_masked


def calculate_aurc(
    y_true,
    probability,
    classification_threshold
):
    y_true = np.asarray(y_true)

    prediction = (
        probability >= classification_threshold
    ).astype("int8")

    confidence = np.maximum(
        probability,
        1 - probability
    )

    errors = (
        prediction != y_true
    ).astype("float64")

    order = np.argsort(-confidence)
    ordered_errors = errors[order]

    cumulative_errors = np.cumsum(
        ordered_errors
    )

    accepted_count = np.arange(
        1,
        len(y_true) + 1
    )

    risks = cumulative_errors / accepted_count

    return float(np.mean(risks))


def evaluate_selective_policy(
    y_true,
    probability,
    classification_threshold,
    confidence_cutoff
):
    y_true = np.asarray(y_true)

    prediction = (
        probability >= classification_threshold
    ).astype("int8")

    confidence = np.maximum(
        probability,
        1 - probability
    )

    accepted = confidence >= confidence_cutoff

    accepted_rows = int(accepted.sum())
    abstained_rows = int((~accepted).sum())

    total_attacks = int((y_true == 1).sum())

    attack_abstentions = int(
        ((~accepted) & (y_true == 1)).sum()
    )

    accepted_false_negatives = int(
        (
            accepted &
            (y_true == 1) &
            (prediction == 0)
        ).sum()
    )

    result = {
        "actual_coverage": float(accepted.mean()),
        "accepted_rows": accepted_rows,
        "abstained_rows": abstained_rows,
        "selective_risk": np.nan,
        "accepted_accuracy": np.nan,
        "accepted_precision": np.nan,
        "accepted_recall": np.nan,
        "accepted_f1": np.nan,
        "attack_abstention_rate":
            attack_abstentions / total_attacks,
        "accepted_false_negative_rate":
            accepted_false_negatives / total_attacks
    }

    if accepted_rows > 0:
        accepted_y = y_true[accepted]
        accepted_prediction = prediction[accepted]

        result.update({
            "selective_risk": float(
                np.mean(
                    accepted_prediction != accepted_y
                )
            ),
            "accepted_accuracy": accuracy_score(
                accepted_y,
                accepted_prediction
            ),
            "accepted_precision": precision_score(
                accepted_y,
                accepted_prediction,
                zero_division=0
            ),
            "accepted_recall": recall_score(
                accepted_y,
                accepted_prediction,
                zero_division=0
            ),
            "accepted_f1": f1_score(
                accepted_y,
                accepted_prediction,
                zero_division=0
            )
        })

    return result
```

2. Define group-aware inputs and non-selective metrics

```python
from sklearn.metrics import (
    balanced_accuracy_score,
    roc_auc_score,
    average_precision_score,
    matthews_corrcoef,
    confusion_matrix,
    brier_score_loss,
    log_loss
)

def create_group_aware_condition(
    X,
    missing_groups
):
    X_condition = np.asarray(
        X,
        dtype=np.float32
    ).copy()

    availability = np.ones(
        (len(X_condition), len(group_names)),
        dtype=np.float32
    )

    for group in missing_groups:
        group_number = group_names.index(group)
        columns = group_indices[group]

        X_condition[:, columns] = (
            training_medians[columns]
        )

        availability[:, group_number] = 0.0

    return np.hstack([
        X_condition,
        availability
    ])


def expected_calibration_error(
    y_true,
    probability,
    bins=15
):
    edges = np.linspace(0, 1, bins + 1)
    ece = 0.0

    for lower, upper in zip(
        edges[:-1],
        edges[1:]
    ):
        if upper == 1:
            selected = (
                (probability >= lower) &
                (probability <= upper)
            )
        else:
            selected = (
                (probability >= lower) &
                (probability < upper)
            )

        count = selected.sum()

        if count == 0:
            continue

        ece += (
            count / len(y_true)
        ) * abs(
            np.mean(y_true[selected])
            - np.mean(probability[selected])
        )

    return float(ece)


def evaluate_nonselective(
    y_true,
    probability,
    threshold
):
    prediction = (
        probability >= threshold
    ).astype("int8")

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        prediction,
        labels=[0, 1]
    ).ravel()

    return {
        "accuracy": accuracy_score(
            y_true, prediction
        ),
        "balanced_accuracy": balanced_accuracy_score(
            y_true, prediction
        ),
        "precision": precision_score(
            y_true, prediction,
            zero_division=0
        ),
        "recall": recall_score(
            y_true, prediction,
            zero_division=0
        ),
        "f1": f1_score(
            y_true, prediction,
            zero_division=0
        ),
        "roc_auc": roc_auc_score(
            y_true, probability
        ),
        "pr_auc": average_precision_score(
            y_true, probability
        ),
        "mcc": matthews_corrcoef(
            y_true, prediction
        ),
        "false_positive_rate":
            fp / (fp + tn),
        "false_negative_rate":
            fn / (fn + tp),
        "brier": brier_score_loss(
            y_true, probability
        ),
        "logloss": log_loss(
            y_true, probability
        ),
        "ece": expected_calibration_error(
            y_true, probability
        ),
        "tn": int(tn),
        "fp": int(fp),
        "fn": int(fn),
        "tp": int(tp)
    }
```

```python
target_nonselective_results = []
target_selective_results = []

conditions = [("complete", tuple())]

conditions.extend([
    ("single", (group,))
    for group in group_names
])

conditions.extend([
    ("pairwise_unseen", pair)
    for pair in combinations(group_names, 2)
])

for condition_type, missing_groups in conditions:
    condition_name = (
        "+".join(missing_groups)
        if missing_groups
        else "None"
    )

    print("\nTesting:", condition_type, condition_name)
    start = time.time()

    # Ordinary baseline
    X_baseline_condition = apply_group_loss(
        X_target_imputed,
        missing_groups,
        replacement="median"
    )

    baseline_raw_probability = (
        baseline_model.predict_proba(
            X_baseline_condition
        )[:, 1]
    )

    baseline_probability = (
        baseline_platt.predict_proba(
            probability_to_logit(
                baseline_raw_probability
            )
        )[:, 1]
    )

    baseline_result = evaluate_nonselective(
        y_target,
        baseline_probability,
        baseline_threshold
    )

    baseline_result.update({
        "model": "ordinary_baseline",
        "condition_type": condition_type,
        "missing_groups": condition_name
    })

    target_nonselective_results.append(
        baseline_result
    )

    baseline_aurc = calculate_aurc(
        y_target,
        baseline_probability,
        baseline_threshold
    )

    for desired_coverage, cutoff in (
        baseline_cutoffs.items()
    ):
        result = evaluate_selective_policy(
            y_target,
            baseline_probability,
            baseline_threshold,
            cutoff
        )

        result.update({
            "model": "ordinary_baseline",
            "condition_type": condition_type,
            "missing_groups": condition_name,
            "desired_policy_coverage":
                desired_coverage,
            "source_confidence_cutoff":
                cutoff,
            "aurc": baseline_aurc
        })

        target_selective_results.append(result)

    del X_baseline_condition
    del baseline_raw_probability
    del baseline_probability
    gc.collect()

    # Group-aware model
    X_group_condition = create_group_aware_condition(
        X_target_imputed,
        missing_groups
    )

    group_raw_probability = (
        group_aware_model.predict_proba(
            X_group_condition
        )[:, 1]
    )

    group_probability = (
        group_aware_platt.predict_proba(
            probability_to_logit(
                group_raw_probability
            )
        )[:, 1]
    )

    group_result = evaluate_nonselective(
        y_target,
        group_probability,
        group_threshold
    )

    group_result.update({
        "model": "group_aware",
        "condition_type": condition_type,
        "missing_groups": condition_name
    })

    target_nonselective_results.append(
        group_result
    )

    group_aurc = calculate_aurc(
        y_target,
        group_probability,
        group_threshold
    )

    for desired_coverage, cutoff in (
        group_cutoffs.items()
    ):
        result = evaluate_selective_policy(
            y_target,
            group_probability,
            group_threshold,
            cutoff
        )

        result.update({
            "model": "group_aware",
            "condition_type": condition_type,
            "missing_groups": condition_name,
            "desired_policy_coverage":
                desired_coverage,
            "source_confidence_cutoff":
                cutoff,
            "aurc": group_aurc
        })

        target_selective_results.append(result)

    del X_group_condition
    del group_raw_probability
    del group_probability
    gc.collect()

    print(
        "Completed in",
        round(time.time() - start, 1),
        "seconds"
    )
```

Step 5: Define fixed augmentation functions

```python
def create_confirmatory_training_data(
    X,
    y,
    seed
):
    X = np.asarray(X, dtype=np.float32)
    y = np.asarray(y, dtype=np.int8)

    number_rows = len(X)
    number_groups = len(group_names)

    rng = np.random.default_rng(seed)

    assigned_groups = rng.integers(
        0,
        number_groups,
        size=number_rows
    )

    X_masked = X.copy()

    masked_availability = np.ones(
        (number_rows, number_groups),
        dtype=np.float32
    )

    assignment_counts = {}

    for group_number, group in enumerate(
        group_names
    ):
        rows = np.where(
            assigned_groups == group_number
        )[0]

        columns = group_indices[group]

        X_masked[np.ix_(rows, columns)] = (
            training_medians[columns]
        )

        masked_availability[
            rows,
            group_number
        ] = 0.0

        assignment_counts[group] = int(
            len(rows)
        )

    complete_availability = np.ones(
        (number_rows, number_groups),
        dtype=np.float32
    )

    X_augmented = np.vstack([
        np.hstack([
            X,
            complete_availability
        ]),
        np.hstack([
            X_masked,
            masked_availability
        ])
    ])

    y_augmented = np.concatenate([
        y,
        y
    ])

    return (
        X_augmented,
        y_augmented,
        assignment_counts
    )


def create_all_known_conditions(X, y):
    X = np.asarray(X, dtype=np.float32)
    y = np.asarray(y, dtype=np.int8)

    X_versions = []
    y_versions = []

    complete_indicators = np.ones(
        (len(X), len(group_names)),
        dtype=np.float32
    )

    X_versions.append(
        np.hstack([
            X,
            complete_indicators
        ])
    )

    y_versions.append(y)

    for group_number, group in enumerate(
        group_names
    ):
        X_masked = X.copy()
        columns = group_indices[group]

        X_masked[:, columns] = (
            training_medians[columns]
        )

        indicators = np.ones(
            (len(X), len(group_names)),
            dtype=np.float32
        )

        indicators[:, group_number] = 0.0

        X_versions.append(
            np.hstack([
                X_masked,
                indicators
            ])
        )

        y_versions.append(y)

    return (
        np.vstack(X_versions),
        np.concatenate(y_versions)
    )
```

1. Define calibration and policy helpers

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    precision_recall_curve,
    brier_score_loss,
    log_loss
)
import numpy as np
import pandas as pd
import joblib
import json
import time
import gc

def probability_to_logit(probability):
    probability = np.clip(
        probability,
        1e-7,
        1 - 1e-7
    )

    return np.log(
        probability / (1 - probability)
    ).reshape(-1, 1)


def expected_calibration_error(
    y_true,
    probability,
    bins=15
):
    y_true = np.asarray(y_true)
    edges = np.linspace(0, 1, bins + 1)
    ece = 0.0

    for lower, upper in zip(
        edges[:-1],
        edges[1:]
    ):
        if upper == 1:
            selected = (
                (probability >= lower) &
                (probability <= upper)
            )
        else:
            selected = (
                (probability >= lower) &
                (probability < upper)
            )

        count = selected.sum()

        if count == 0:
            continue

        ece += (
            count / len(y_true)
        ) * abs(
            np.mean(y_true[selected])
            - np.mean(probability[selected])
        )

    return float(ece)


def fit_platt_and_policy(
    model,
    X_calibrator,
    y_calibrator,
    X_policy,
    y_policy,
    seed
):
    raw_calibrator_probability = (
        model.predict_proba(
            X_calibrator
        )[:, 1]
    )

    calibrator = LogisticRegression(
        C=1_000_000,
        solver="lbfgs",
        max_iter=1000,
        random_state=seed
    )

    calibrator.fit(
        probability_to_logit(
            raw_calibrator_probability
        ),
        y_calibrator
    )

    raw_policy_probability = (
        model.predict_proba(
            X_policy
        )[:, 1]
    )

    calibrated_policy_probability = (
        calibrator.predict_proba(
            probability_to_logit(
                raw_policy_probability
            )
        )[:, 1]
    )

    precision_values, recall_values, thresholds = (
        precision_recall_curve(
            y_policy,
            calibrated_policy_probability
        )
    )

    f1_values = (
        2
        * precision_values[:-1]
        * recall_values[:-1]
        /
        (
            precision_values[:-1]
            + recall_values[:-1]
            + 1e-12
        )
    )

    best_position = int(
        np.argmax(f1_values)
    )

    classification_threshold = float(
        thresholds[best_position]
    )

    policy_confidence = np.maximum(
        calibrated_policy_probability,
        1 - calibrated_policy_probability
    )

    confidence_cutoffs = {
        str(coverage): float(
            np.quantile(
                policy_confidence,
                1 - coverage
            )
        )
        for coverage in [
            0.80,
            0.90,
            0.95
        ]
    }

    calibration_metrics = {
        "raw": {
            "brier": float(
                brier_score_loss(
                    y_policy,
                    raw_policy_probability
                )
            ),
            "logloss": float(
                log_loss(
                    y_policy,
                    raw_policy_probability
                )
            ),
            "ece": expected_calibration_error(
                y_policy,
                raw_policy_probability
            )
        },
        "platt_calibrated": {
            "brier": float(
                brier_score_loss(
                    y_policy,
                    calibrated_policy_probability
                )
            ),
            "logloss": float(
                log_loss(
                    y_policy,
                    calibrated_policy_probability
                )
            ),
            "ece": expected_calibration_error(
                y_policy,
                calibrated_policy_probability
            )
        }
    }

    policy = {
        "classification_threshold":
            classification_threshold,
        "policy_f1": float(
            f1_values[best_position]
        ),
        "confidence_cutoffs":
            confidence_cutoffs,
        "calibration_method":
            "Platt scaling",
        "seed": seed,
        "target_data_used": False
    }

    return (
        calibrator,
        policy,
        calibration_metrics
    )
```

2. Define the fixed XGBoost configuration

```python
def create_fixed_xgboost(seed):
    return xgb.XGBClassifier(
        objective="binary:logistic",
        n_estimators=1200,
        learning_rate=0.05,
        max_depth=6,
        min_child_weight=5,
        subsample=0.80,
        colsample_bytree=0.80,
        reg_alpha=0.01,
        reg_lambda=1.0,
        tree_method="hist",
        device="cpu",
        eval_metric="logloss",
        early_stopping_rounds=50,
        random_state=seed,
        n_jobs=-1
    )
```

3. Define the restartable seed-training function

```python
def train_confirmatory_seed(seed):
    seed_dir = (
        replication_dir /
        f"seed_{seed}"
    )

    seed_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    completion_file = (
        seed_dir /
        "TRAINING_COMPLETE.json"
    )

    if completion_file.exists():
        print(
            f"Seed {seed} is already complete."
        )
        return

    summary = {
        "seed": seed,
        "target_data_used": False
    }

    print("\n" + "=" * 70)
    print("CONFIRMATORY SEED:", seed)
    print("=" * 70)

    # ------------------------------------------------
    # Ordinary baseline
    # ------------------------------------------------
    print("\nTraining ordinary baseline")

    baseline_start = time.time()

    baseline_model_seed = (
        create_fixed_xgboost(seed)
    )

    baseline_model_seed.fit(
        X_train_fixed,
        y_train_fixed,
        eval_set=[
            (
                X_validation_fixed,
                y_validation_fixed
            )
        ],
        verbose=50
    )

    baseline_time = (
        time.time() - baseline_start
    )

    baseline_model_seed.save_model(
        str(
            seed_dir /
            "ordinary_baseline_xgboost.json"
        )
    )

    (
        baseline_calibrator_seed,
        baseline_policy_seed,
        baseline_calibration_metrics
    ) = fit_platt_and_policy(
        model=baseline_model_seed,
        X_calibrator=X_calibrator_base,
        y_calibrator=y_calibrator_base,
        X_policy=X_policy_base,
        y_policy=y_policy_base,
        seed=seed
    )

    joblib.dump(
        baseline_calibrator_seed,
        seed_dir /
        "ordinary_baseline_platt.joblib"
    )

    with open(
        seed_dir /
        "ordinary_baseline_policy.json",
        "w"
    ) as file:
        json.dump(
            baseline_policy_seed,
            file,
            indent=2
        )

    with open(
        seed_dir /
        "ordinary_baseline_calibration.json",
        "w"
    ) as file:
        json.dump(
            baseline_calibration_metrics,
            file,
            indent=2
        )

    summary["ordinary_baseline"] = {
        "training_time_seconds":
            baseline_time,
        "best_iteration": int(
            baseline_model_seed.best_iteration
        ),
        "best_validation_logloss": float(
            baseline_model_seed.best_score
        ),
        "classification_threshold":
            baseline_policy_seed[
                "classification_threshold"
            ],
        "policy_f1":
            baseline_policy_seed[
                "policy_f1"
            ]
    }

    print(
        "Baseline completed in",
        round(baseline_time, 1),
        "seconds"
    )

    # ------------------------------------------------
    # Group-aware model
    # ------------------------------------------------
    print("\nCreating group-aware training data")

    (
        X_group_train,
        y_group_train,
        assignment_counts
    ) = create_confirmatory_training_data(
        X_train_fixed,
        y_train_fixed,
        seed
    )

    print(
        "Group-aware training shape:",
        X_group_train.shape
    )

    print(
        "Mask assignment counts:",
        assignment_counts
    )

    print("\nTraining group-aware model")

    group_start = time.time()

    group_model_seed = (
        create_fixed_xgboost(seed)
    )

    group_model_seed.fit(
        X_group_train,
        y_group_train,
        eval_set=[
            (
                X_group_validation,
                y_group_validation
            )
        ],
        verbose=50
    )

    group_time = (
        time.time() - group_start
    )

    group_model_seed.save_model(
        str(
            seed_dir /
            "group_aware_xgboost.json"
        )
    )

    del X_group_train
    del y_group_train
    gc.collect()

    (
        group_calibrator_seed,
        group_policy_seed,
        group_calibration_metrics
    ) = fit_platt_and_policy(
        model=group_model_seed,
        X_calibrator=X_group_calibrator,
        y_calibrator=y_group_calibrator,
        X_policy=X_group_policy,
        y_policy=y_group_policy,
        seed=seed
    )

    joblib.dump(
        group_calibrator_seed,
        seed_dir /
        "group_aware_platt.joblib"
    )

    with open(
        seed_dir /
        "group_aware_policy.json",
        "w"
    ) as file:
        json.dump(
            group_policy_seed,
            file,
            indent=2
        )

    with open(
        seed_dir /
        "group_aware_calibration.json",
        "w"
    ) as file:
        json.dump(
            group_calibration_metrics,
            file,
            indent=2
        )

    summary["group_aware"] = {
        "training_time_seconds":
            group_time,
        "best_iteration": int(
            group_model_seed.best_iteration
        ),
        "best_validation_logloss": float(
            group_model_seed.best_score
        ),
        "classification_threshold":
            group_policy_seed[
                "classification_threshold"
            ],
        "policy_f1":
            group_policy_seed[
                "policy_f1"
            ],
        "single_mask_assignment_counts":
            assignment_counts
    }

    with open(
        seed_dir /
        "training_summary.json",
        "w"
    ) as file:
        json.dump(
            summary,
            file,
            indent=2
        )

    with open(
        completion_file,
        "w"
    ) as file:
        json.dump(
            {
                "seed": seed,
                "status":
                    "training_complete",
                "target_data_used": False
            },
            file,
            indent=2
        )

    print(
        "Group-aware model completed in",
        round(group_time, 1),
        "seconds"
    )

    print(
        f"\nSeed {seed} training and "
        "source-only policy creation complete."
    )

    del baseline_model_seed
    del baseline_calibrator_seed
    del group_model_seed
    del group_calibrator_seed

    gc.collect()
```
